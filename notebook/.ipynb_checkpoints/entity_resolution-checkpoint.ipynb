```python
import sys, os
sys.path.append(os.path.abspath("../code/business_entity_resolution/src"))

import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import GroupShuffleSplit

from blocking import generate_candidates
from features import build_pair_features, ALL_FEATURE_COLUMNS
from metrics import macro_f_beta, sweep_threshold

DATA_DIR = "../dataset/train"
```

```python
source1 = pd.read_csv(f"{DATA_DIR}/train_source1.tsv", sep="\t", dtype=str)
source2 = pd.read_csv(f"{DATA_DIR}/train_source2.tsv", sep="\t", dtype=str)
source3 = pd.read_csv(f"{DATA_DIR}/train_source3.tsv", sep="\t", dtype=str)

def load_ground_truth(path):
    gt = pd.read_csv(path, sep="\t", dtype=str, keep_default_na=False)
    return {
        row["source1_entity_id"]: set(row["matched_entity_ids"].split(",")) if row["matched_entity_ids"] else set()
        for _, row in gt.iterrows()
    }

ground_truth = load_ground_truth(f"{DATA_DIR}/train_ground_truth.tsv")

print(f"S1={len(source1)}  S2={len(source2)}  S3={len(source3)}  GT rows={len(ground_truth)}")
source1.head()
```

```python
candidates = generate_candidates(source1, source2, source3, top_k=30)
print(f"Candidate pairs: {len(candidates)}")
print(f"Avg candidates per S1 entity: {len(candidates) / source1['entity_id'].nunique():.1f}")

true_pairs = {(s1, m) for s1, ms in ground_truth.items() for m in ms}
cand_pairs = set(zip(candidates["source1_entity_id"], candidates["candidate_entity_id"]))
recall_ceiling = len(true_pairs & cand_pairs) / max(len(true_pairs), 1)
print(f"Blocking recall ceiling: {recall_ceiling:.4f}")
```

```python
records = pd.concat([source1, source2, source3], ignore_index=True).set_index("entity_id")
feat = build_pair_features(candidates, records)

def is_match(row):
    return int(row["candidate_entity_id"] in ground_truth.get(row["source1_entity_id"], set()))

feat["label"] = feat.apply(is_match, axis=1)
print(f"Positive rate: {feat['label'].mean():.4f}")
feat[ALL_FEATURE_COLUMNS + ["label"]].describe()
```

```python
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, val_idx = next(splitter.split(feat, groups=feat["source1_entity_id"]))
train_df, val_df = feat.iloc[train_idx].copy(), feat.iloc[val_idx].copy()

X_train, y_train = train_df[ALL_FEATURE_COLUMNS], train_df["label"]
X_val, y_val = val_df[ALL_FEATURE_COLUMNS], val_df["label"]
print(f"Train pairs: {len(X_train)}  Val pairs: {len(X_val)}")
```

```python
pos_weight = (y_train == 0).sum() / max((y_train == 1).sum(), 1)

model = xgb.XGBClassifier(
    n_estimators=400, max_depth=6, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8,
    scale_pos_weight=pos_weight, eval_metric="aucpr", n_jobs=-1,
)
model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=True)

# quick feature importance check
pd.Series(model.feature_importances_, index=ALL_FEATURE_COLUMNS).sort_values(ascending=False)
```

```python
val_df["score"] = model.predict_proba(X_val)[:, 1]

val_entities = set(val_df["source1_entity_id"])
val_gt = {k: v for k, v in ground_truth.items() if k in val_entities}
for eid in val_entities - set(val_gt):
    val_gt[eid] = set()

best_t, best_f05, sweep_results = sweep_threshold(val_df, val_gt, margin_min=0.0)
print(f"Best threshold={best_t:.2f}  Validation macro F0.5={best_f05:.4f}")

pd.DataFrame(sweep_results, columns=["threshold", "f05"]).plot(x="threshold", y="f05", figsize=(6,4))
```

```python
import json
MODEL_PATH = "../model.json"
model.save_model(MODEL_PATH)
with open(MODEL_PATH + ".meta.json", "w") as f:
    json.dump({"threshold": best_t, "margin_min": 0.0, "features": ALL_FEATURE_COLUMNS}, f, indent=2)
print("Saved:", MODEL_PATH)
```

```python
TEST_DIR = "../dataset/test"
OUT_DIR = "../output"

t_source1 = pd.read_csv(f"{TEST_DIR}/test_source1.tsv", sep="\t", dtype=str)
t_source2 = pd.read_csv(f"{TEST_DIR}/test_source2.tsv", sep="\t", dtype=str)
t_source3 = pd.read_csv(f"{TEST_DIR}/test_source3.tsv", sep="\t", dtype=str)

test_candidates = generate_candidates(t_source1, t_source2, t_source3, top_k=30)

cand_out = (
    test_candidates.dropna(subset=["candidate_entity_id"])
    .groupby("source1_entity_id")["candidate_entity_id"]
    .apply(lambda s: ",".join(s))
    .reindex(t_source1["entity_id"].rename("source1_entity_id"), fill_value="")
    .reset_index()
    .rename(columns={"candidate_entity_id": "candidate_entity_ids"})
)
cand_out.to_csv(f"{OUT_DIR}/candidate_pairs.tsv", sep="\t", index=False)

t_records = pd.concat([t_source1, t_source2, t_source3], ignore_index=True).set_index("entity_id")
test_feat = build_pair_features(test_candidates, t_records)
test_feat["score"] = model.predict_proba(test_feat[ALL_FEATURE_COLUMNS])[:, 1]

keep = test_feat[test_feat["score"] >= best_t]
match_out = (
    keep.groupby("source1_entity_id")["candidate_entity_id"]
    .apply(lambda s: ",".join(sorted(set(s))))
    .reindex(t_source1["entity_id"].rename("source1_entity_id"), fill_value="")
    .reset_index()
    .rename(columns={"candidate_entity_id": "matched_entity_ids"})
)
match_out.to_csv(f"{OUT_DIR}/matching_results.tsv", sep="\t", index=False)
print("Wrote candidate_pairs.tsv and matching_results.tsv")
match_out.head()
```

```python
import subprocess
result = subprocess.run(
    ["python", "../utils/validate_submission.py",
     "--matching", "../output/matching_results.tsv",
     "--candidate", "../output/candidate_pairs.tsv",
     "--test-dir", "../dataset/test"],
    capture_output=True, text=True
)
print(result.stdout)
print(result.stderr)
```
